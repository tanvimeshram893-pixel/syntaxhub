```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
data=pd.read_csv("/content/drive/MyDrive/Salary Data.csv")
```

```python
data.head()
```

```python
X=data.iloc[:,4].values
y=data.iloc[:,5].values
```

```python
X
```

```python
y
```

```python
X.shape
```

```python
y.shape
```

```python
X=X.reshape(-1,1)
y=y.reshape(-1,1)

from sklearn.impute import SimpleImputer

imputer_X = SimpleImputer(missing_values=np.nan, strategy='mean')
X = imputer_X.fit_transform(X)

imputer_y = SimpleImputer(missing_values=np.nan, strategy='mean')
y = imputer_y.fit_transform(y)
```

```python
X
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split
Xtrain,Xtest,ytrain,ytest=train_test_split(X,y,test_size=0.2)
```

```python
Xtrain
```

```python
Xtrain.shape
```

```python
Xtest.shape
```

```python
from sklearn.linear_model import LinearRegression
lin=LinearRegression()
```

```python
lin.fit(Xtrain,ytrain)
```

```python
y_pred=lin.predict(Xtest)
```

```python
y_pred[5]
```

```python
ytest[5]
```

```python
from sklearn.metrics import r2_score
r2_score(ytest,y_pred)
```

```python
plt.scatter(Xtrain,ytrain,color='red')
plt.plot(Xtrain,lin.predict(Xtrain),color='blue')
```

```python
plt.scatter(Xtrain,ytrain,color='red')
plt.plot(Xtrain,lin.predict(Xtrain),color='blue')
plt.xlabel("years of experience")
plt.ylabel("salary")
plt.show()
```

```python
plt.scatter(Xtest,ytest,color='red')
plt.plot(Xtest,lin.predict(Xtest),color='blue')
plt.xlabel("years of experience")
plt.ylabel("salary")
plt.show()
```
